```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
popgames100 = pd.read_csv("./Presentation/games.csv")
games2024 = pd.read_csv("./Presentation/vgchartz-2024.csv")
top100sales = pd.read_csv("./Presentation/vgsales.csv")
steamrec = pd.read_csv("./Presentation/games_oth.csv")
```

```python
global_sales = top100sales.groupby('Genre')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum()

global_sales[['Global_Sales']].\
    sort_values(by='Global_Sales', ascending=False).\
    plot(kind='bar', figsize=(14,8), color='black')

plt.title('Global sales per genre')
plt.xlabel('Genre')
plt.ylabel('Sales (millions)')
plt.show()
```

```python
region_genre = top100sales.groupby('Genre', as_index=False)[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales']].sum()

region_genre.plot(x='Genre', y=['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales'], kind='bar', figsize=(14,8))
plt.title('Sales in each region based on Genre')
plt.xlabel('Genre')
plt.ylabel('Sales (millions)')
plt.grid(axis='y')
plt.legend(["America", "EU", "Japan", "Other"])
plt.show()
```

```python
platform_sales = top100sales.groupby('Platform')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum()

platform_sales[['Global_Sales']].\
    sort_values(by='Global_Sales', ascending=False).\
    plot(kind='bar', figsize=(14,8), color='black')

plt.title('Global sales')
plt.xlabel('Platform')
plt.ylabel('Sales (millions)')
plt.show()
```

```python
publisher_sales = top100sales.groupby('Publisher')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum().nlargest(10, 'Global_Sales')

publisher_sales[['Global_Sales']].sort_values(by='Global_Sales', ascending=False).plot(kind='bar', figsize=(14,8), color='black')
plt.title('Global sales')
plt.xlabel('Publisher')
plt.ylabel('Sales (millions)')
plt.show()
```

```python
countplat = top100sales.groupby(['Name'])['Name'].count()
mostplat = countplat.nlargest(20)
```

```python
mostplat_rows = top100sales[top100sales['Name'].isin(mostplat.index)]
```

```python
mostplat_sales = mostplat_rows.groupby('Name')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum().nlargest(10, 'Global_Sales')
mostplat_sales[['Global_Sales']].sort_values(by='Global_Sales', ascending=False).plot(kind='bar', figsize=(14,8), color='black')
plt.title('Global sales')
plt.xlabel('Name')
plt.ylabel('Sales (millions)')
plt.show()
```

```python
top10sales = top100sales[0:10].groupby('Name')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum()
top10sales[['Global_Sales']].sort_values(by='Global_Sales', ascending=False).plot(kind='bar', figsize=(14,8), color='black')
plt.title('Global sales')
plt.xlabel('Name')
plt.ylabel('Sales (millions)')
plt.show()
```

```python
top10oth = oth_platforms['Publisher'].value_counts().head(5)
alltop = pd.concat([nintendo, oth_platforms[oth_platforms['Publisher'].isin(top10oth.index)]], ignore_index=True)
alltop_plt = alltop.groupby(['Publisher','Year']).size().reset_index(name='counts')
plt.figure(figsize=(14,8))
sns.lineplot(data=alltop_plt, x = 'Year', y = 'counts', hue = 'Publisher', palette = 'tab10')
plt.title('Release of games per year by publisher')
plt.xlabel('Year')
plt.ylabel('Number of games')
plt.xticks(range(1980,2017), rotation=90)
plt.grid(axis='x')
plt.show()
```

```python
alltop_plt = alltop.groupby(['Publisher','Year']).sum('Global_Sales')
plt.figure(figsize=(14,8))
sns.lineplot(data=alltop_plt, x = 'Year', y = 'Global_Sales', hue = 'Publisher', palette = 'tab10')
plt.title('Sales per year by publisher')
plt.xlabel('Year')
plt.ylabel('Sales (millions)')
plt.xticks(range(1980,2017), rotation=90)
plt.grid(axis='x')
plt.show()
```

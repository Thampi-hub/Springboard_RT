```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
top100sales = pd.read_csv("./vgsales.csv")
top100sales.head()
```

```python
top100sales['Global_Sales'].describe()
```

```python
top100sales.info()
```

```python
mil_above_cnt = top100sales[top100sales['Global_Sales']>1].shape[0]/top100sales.shape[0] * 100
print(f"Games which made more than 1 million = {round(mil_above_cnt, 2)}%")
```

```python
plt.figure(figsize=(40,2))
plt.boxplot(top100sales['Global_Sales'],vert=0)
plt.title('Global sales')
plt.xlabel('Name')
plt.ylabel('Sales (millions)')
plt.show()
```

```python
top100sales.nlargest(10, 'NA_Sales')
```

```python
top100sales.nlargest(10, 'EU_Sales')
```

```python
top100sales.nlargest(10, 'Global_Sales')
```

```python
publisher_sales = top100sales.groupby('Publisher')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum().nlargest(10, 'Global_Sales')

publisher_sales[['Global_Sales']].sort_values(by='Global_Sales', ascending=False).plot(kind='bar', figsize=(14,8), color='cadetblue')
plt.title('Global sales')
plt.xlabel('Publisher')
plt.ylabel('Sales (millions)')
plt.axhline(y = 300, color = 'r', linestyle = 'dashed') 
plt.show()
```

```python
nintendo = top100sales[top100sales['Publisher']=='Nintendo']
platform_sales = nintendo.groupby('Platform')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum()
platform_sales.sort_values('Global_Sales', ascending=False)
```

```python
nintendo.nsmallest(10,'Rank')
```

```python
top10sales = top100sales[0:20].groupby('Name')[['NA_Sales', 'EU_Sales', 'JP_Sales', 'Other_Sales', 'Global_Sales']].sum()
top10sales[['Global_Sales']].sort_values(by='Global_Sales', ascending=False).plot(kind='bar', figsize=(14,8), color='cadetblue')
plt.title('Global sales')
plt.xlabel('Game')
plt.ylabel('Sales (millions)')
plt.grid(axis='y')
plt.show()
```

```python
top100sales[['Name','Publisher','Platform']][0:20]
```

```python
oth_platforms = top100sales[top100sales['Publisher'].isin(publisher_sales[1:7].index)]
top10oth = oth_platforms['Publisher'].value_counts().head(6)
alltop = pd.concat([nintendo, oth_platforms[oth_platforms['Publisher'].isin(top10oth.index)]], ignore_index=True)
alltop_plt = alltop.groupby(['Publisher','Year']).size().reset_index(name='counts')
plt.figure(figsize=(14,8))
sns.lineplot(data=alltop_plt, x = 'Year', y = 'counts', hue = 'Publisher', palette = 'Set2')
plt.title('Release of games per year by publisher')
plt.xlabel('Year')
plt.ylabel('Number of games')
plt.xticks(range(1980,2021), rotation=90)
plt.grid(axis='x')
plt.xlim(1979,2021)
plt.show()
```

```python
alltop_plt = alltop.groupby(['Publisher','Year']).sum('Global_Sales')
plt.figure(figsize=(14,8))
sns.lineplot(data=alltop_plt, x = 'Year', y = 'Global_Sales', hue = 'Publisher', palette = 'Set2')
plt.title('Sales per year by publisher')
plt.xlabel('Year')
plt.ylabel('Sales (millions)')
plt.xticks(range(1980,2021), rotation=90)
plt.grid(axis='x')
plt.xlim(1979,2021)
plt.show()
```
